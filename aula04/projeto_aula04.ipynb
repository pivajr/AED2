### Busca Eficiente em um Índice de Livros   
**Contexto:** Você está trabalhando em um projeto para gerenciar uma biblioteca. As informações sobre os livros (título, autor, ano de publicação) estão armazenadas em uma lista, e essa lista é mantida ordenada alfabeticamente pelo título do livro.   
**Problema:** Os usuários precisam encontrar rapidamente um livro específico sabendo seu título. Uma busca ineficiente pode dificultar a localização dos livros desejados.   


```python
def busca_livro(lista_livros, titulo):
    """
    Realiza busca binária em uma lista de tuplas (titulo, autor, ano).
    Retorna a tupla correspondente se encontrar, ou None caso contrário.
    """
    inicio = 0
    fim = len(lista_livros) - 1
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        titulo_meio = lista_livros[meio][0]
        
        if titulo_meio == titulo:
            return lista_livros[meio]
        elif titulo < titulo_meio:
            fim = meio - 1
        else:
            inicio = meio + 1
    
    return None
```

```python
# Lista de livros ordenada alfabeticamente pelo título
livros = [
    ("A Arte da Guerra", "Sun Tzu", -500),
    ("Cem Anos de Solidão", "Gabriel García Márquez", 1967),
    ("Dom Casmurro", "Machado de Assis", 1899),
    ("O Hobbit", "J.R.R. Tolkien", 1937),
    ("Orgulho e Preconceito", "Jane Austen", 1813)
]
```

### Testando a Implementação

```python
# Buscando livros
print(busca_livro(livros, "Dom Casmurro"))  
```

```python
def busca_produto(lista_produtos, nome_produto):
    """
    Realiza busca binária em uma lista de dicionários (nome, preço e descrição).
    Retorna o dicionário correspondente se encontrar, ou None caso contrário.
    """
    
    inicio = 0
    fim = len(lista_produtos) - 1
    
    # Normalizamos o termo de busca para minúsculo e removemos espaços extras
    nome_busca = nome_produto.strip().lower()
    
    while inicio <= fim:
        meio = (inicio + fim) // 2
        # Normalizamos o nome da lista para a comparação
        nome_meio = lista_produtos[meio]["nome"].strip().lower()
        
        if nome_meio == nome_busca:
            return lista_produtos[meio]
        elif nome_busca < nome_meio:
            fim = meio - 1
        else:
            inicio = meio + 1
    
    return None
```
