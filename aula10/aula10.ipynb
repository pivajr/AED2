```python
def soma_digitos(n):
    if n < 10:  # Caso base: número com um único dígito
        return n
    else:
        ultimo_digito = n % 10
        resto_do_numero = n // 10
        return ultimo_digito + soma_digitos(resto_do_numero)  # Passo recursivo
```

```python
print(soma_digitos(12345))
```

```python
def soma_digitos2(n):
    tamanho = len(n)
    if tamanho > 1:
        v = int(n[0])
        nova = ''
        for i in range(1, tamanho):
            nova = nova + n[i]
        return v + soma_digitos2(nova)
    else:
        return int(n)
```

```python
print(soma_digitos2('12345'))
```
